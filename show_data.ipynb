```python
path = os.getcwd()
print(f"current path: {path}")
```

```python
dst_path = os.path.join(path, "data_dst")
```

```python
df_matriculados_T = pd.read_csv(os.path.join(dst_path, "matriculados_T_STEM.csv"))
df_graduados_T = pd.read_csv(os.path.join(dst_path, "graduados_T_STEM.csv"))
df_matriculados_stats = pd.read_csv(os.path.join(dst_path, "matriculados_T_STEM_stats.csv"))
df_graduados_stats = pd.read_csv(os.path.join(dst_path, "graduados_T_STEM_stats.csv"))
```

```python
df_graduados_T.head()
```

```python
df_matriculados_stats[["Titulacion", "Media_Porcentaje_Mujeres", "Media_Porcentaje_Hombres"]].sort_values(by="Media_Porcentaje_Mujeres",ascending=True)
```

```python
df_graduados_stats[["Titulacion", "Media_Porcentaje_Mujeres", "Media_Porcentaje_Hombres"]].sort_values(by="Media_Porcentaje_Mujeres",ascending=True)
```

```python
df_diferencia = pd.merge(
    df_matriculados_stats[
        ["Titulacion", "Media_Porcentaje_Mujeres"]
    ],
    df_graduados_stats[
        ["Titulacion", "Media_Porcentaje_Mujeres"]
    ],
    on="Titulacion",
    how="inner",
    suffixes=("_Matriculadas", "_Graduadas")
)

df_diferencia["Diferencia_Mujeres_pp"] = (
    df_diferencia["Media_Porcentaje_Mujeres_Graduadas"]
    - df_diferencia["Media_Porcentaje_Mujeres_Matriculadas"]
).round(2)

df_diferencia.sort_values("Diferencia_Mujeres_pp")
```
